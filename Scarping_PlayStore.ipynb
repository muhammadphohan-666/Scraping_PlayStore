```python
from google_play_scraper import app

APP_ID = "id.co.bri.brimo"  # coba ID kamu dulu

try:
    info = app(APP_ID, lang="id", country="id")
    print("✅ App ditemukan!")
    print("Title:", info.get("title"))
    print("AppId:", info.get("appId"))
except Exception as e:
    print("❌ AppId tidak valid / tidak ditemukan:", e)
```

```python
from google_play_scraper import reviews, Sort
import pandas as pd

APP_ID = "com.bca"     # ✅ BCA mobile
COUNT = 2000
LANG = "id"
COUNTRY = "id"

result, _ = reviews(
    APP_ID,
    lang=LANG,
    country=COUNTRY,
    sort=Sort.NEWEST,
    count=COUNT
)

df = pd.DataFrame(result)
print("Jumlah review keambil:", len(df))

df.to_csv("bca_mobile_reviews.csv", index=False, encoding="utf-8-sig")
print("Selesai! File: bca_mobile_reviews.csv")
```
